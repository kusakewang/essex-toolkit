## Embeddings

Sample notebook to showcase embeddings usage.

Required imports:

```python
# Copyright (c) 2025 Microsoft Corporation.

import logging
import os

from fnllm.events.logger import LLMEventsLogger
from fnllm.openai import AzureOpenAIConfig, create_openai_embeddings_llm, create_openai_embeddings_rest_llm
from dotenv import load_dotenv

load_dotenv()
```

Example logger to showcase event logging:

```python
def _create_logger():
    logger = logging.getLogger("CHAT")
    logger.setLevel(logging.DEBUG)

    handler = logging.StreamHandler()
    handler.setLevel(logging.DEBUG)
    handler.setFormatter(
        logging.Formatter("[%(levelname)s %(asctime)s %(name)s] - %(message)s")
    )
    logger.handlers = [handler]

    return logger
```

Creating embeddings client:

```python
config = AzureOpenAIConfig(
    endpoint=os.environ["OPENAI_API_ENDPOINT"],
    api_version=os.environ["OPENAI_API_VERSION"],
    model=os.environ["OPENAI_API_MODEL"],
)

embeddings_client = create_openai_embeddings_llm(
    config,
    events=LLMEventsLogger(),
)
```

```python
res = await embeddings_client("Embed this for me")
```

```python
print(res.model_dump_json(indent=2))
```

```python
BASE_URL = "https://data-ai-dev.microsoft.com"
API_KEY = os.environ["API_KEY"]
MODEL = "metisaca"
# Create the REST-based embeddings LLM
llm = create_openai_embeddings_rest_llm(
    base_url=BASE_URL,
    api_key=API_KEY,
    model=MODEL
)

result = await llm("Embed this for me")
print(result)
```
